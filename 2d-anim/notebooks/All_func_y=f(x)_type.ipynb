```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML


def setup_axes():
    fig, ax = plt.subplots(figsize=(6, 6), dpi=100, facecolor="cyan", edgecolor="black", linewidth=3)
    ax.set_xlim(-5, 5)
    ax.set_ylim(-5, 5)
    ax.axhline(0, color="black", linewidth=2, linestyle=":")
    ax.axvline(0, color="black", linewidth=2, linestyle=":")
    ax.set_aspect("equal", adjustable="box")
    ax.grid(True)
    ax.set_title("Finding Locus for Various Functions")
    ax.set_xlabel("x-axis")
    ax.set_ylabel("y-axis")
    return fig, ax


def plot_function(ax, set_of_points):
    ax.plot(set_of_points, [2] * len(set_of_points), linewidth=2, color="red", label="y=2")
    ax.legend()


def starting_position():
    point.set_data([], [])
    return point,


def update_frame(frame):
    point.set_data([frame], [2])
    return point,


set_of_points = np.linspace(-6, 6, 100)
figure, ax = setup_axes()
plot_function(ax, set_of_points)
point, = ax.plot([], [], "bo", markersize=10)
plt.close(figure)

ani = animation.FuncAnimation(
    fig=figure,
    func=update_frame,
    frames=set_of_points,
    init_func=starting_position,
    interval=50,
    blit=True
)

HTML(ani.to_jshtml())
```
